```python
from sklearn.datasets import fetch_20newsgroups
```

```python
twenty_train = fetch_20newsgroups(subset="train", shuffle=True)
```

```python
twenty_train.keys()
```

```python
twenty_train["data"][0]
```

```python
twenty_train.target_names
```

```python
twenty_train.target
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
count_vect = CountVectorizer()
```

```python
X_train_counts = count_vect.fit_transform(twenty_train["data"])
X_train_counts.shape
```

```python
print(X_train_counts[0])
```

```python
from sklearn.feature_extraction.text import TfidfTransformer
```

```python
tfidf_transformer = TfidfTransformer()
```

```python
X_train_tfidf = tfidf_transformer.fit_transform(X_train_counts)
```

```python
from sklearn.svm import LinearSVC # svm support vector machine
```

```python

clf_svc = LinearSVC(
    # hyperparameters
    penalty="l2", # l2 regularization, minimalizacja overfitingu
    dual= "auto", # rozwiazuje problem dual, if more features than samples
    tol=0.001, # tolerancja,
)
```

```python
clf_svc.fit(X_train_tfidf, twenty_train.target)
```
