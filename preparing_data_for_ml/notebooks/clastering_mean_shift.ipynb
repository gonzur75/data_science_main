```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../datasets/titanic.csv")
```

```python
df.head()
```

```python
df.drop(columns=["PassengerId", "Name", "Ticket", "Cabin"], inplace=True)
```

```python
df.shape
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()

df["Sex"] = le.fit_transform(df["Sex"]).astype(int)
```

```python
df["Sex"]
```

```python
df = pd.get_dummies(df, columns=["Embarked"], dtype=np.float64)
```

```python
df.shape
```

```python
df.loc[df.isnull().any(axis=1)]
```

```python
df.drop(columns=["Age"], inplace=True)
```

```python
df.loc[df.isnull().any(axis=1)]
```

```python
from sklearn.cluster import MeanShift, estimate_bandwidth
```

```python

model = MeanShift(
    bandwidth=30,
    n_jobs=4 # ile procesorów ma użyć
)
model.fit(df)
```

```python
estimate_bandwidth(df)
```

```python
model.labels_
```

```python
np.unique(model.labels_)
```

```python
df["cluster_group"] = model.labels_
```

```python
df.head()
```

```python
df["cluster_group"].value_counts()
```

```python
df_data = df.groupby(["cluster_group"]).mean()
```
