```python
from pandas._typing import FilePath, ReadCsvBuffer

type CSVBuffer = FilePath | ReadCsvBuffer[bytes] | ReadCsvBuffer[str]
```

```python
def check_columns(file_path_of_buffer: CSVBuffer, cols_mapper: dict[str, str | pd.Categorical], skiprows: int = 0) -> set[str]:
    cols = pd.read_csv(file_path_of_buffer, skiprows=skiprows, nrows=0).columns.tolist()
    cols_set = set(cols)
    cols_mapper_set = set(cols_mapper)
    return cols_set - cols_mapper_set
```

```python
file_path = "https://github.com/jonfernandes/pandas_essential/raw/main/olympics_1896_2004.csv"
```

```python
check_columns(file_path, dtype_mapper, skiprows=5)
```
